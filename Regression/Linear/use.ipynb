```python
import torch
from Linear import LinearRegression
import matplotlib.pyplot as plt
```

```python
def gen_2D_dataset(num=100):
    X = torch.rand(num,1)   #(num,1)
    y = 3 * X + 2           #(num,1)
    
    y += torch.randn(num,1) * 0.2
    
    return X, y

X_train, y_train = gen_2D_dataset(100)
X_test, y_test = gen_2D_dataset(10) 
```

# Linear Regression

```python
linear=LinearRegression()
linear.closed_form_solution(X_train,y_train)

print(linear.predict(X_test))
print(linear.w_hat)
```

```python
X_range = torch.tensor([[0.0], [1.0]]) 
y_range = linear.predict(X_range)

plt.figure(figsize=(10, 6))
plt.scatter(X_train.numpy(), y_train.numpy(), color='blue', alpha=0.5, label='Training Data')
plt.scatter(X_test.numpy(), y_test.numpy(), color='red', marker='x', s=100, label='Test Data')

plt.plot(X_range.numpy(), y_range.numpy(), color='green', linewidth=3, label='Regression Line')

plt.title('Linear Regression Visualization', fontsize=15)
plt.xlabel('X (Input)', fontsize=12)
plt.ylabel('y (Output)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```
