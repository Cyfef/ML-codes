```python
import torch
from Linear import Linear_Regression
```

```python
def gen_2D_dataset(num=100):
    X = torch.rand(num,1)   #(num,1)
    y = 3 * X + 2           #(num,1)
    
    y += torch.randn(num,1) * 0.2
    
    return X, y

X_train, y_train = gen_2D_dataset(100)
X_test, y_test = gen_2D_dataset(10) 
```

# Linear Regression

```python
LinearRegression=Linear_Regression()
LinearRegression.closed_form_solution(X_train,y_train)

print(LinearRegression.predict(X_test))
print(LinearRegression.w_hat)
```
