```python
import torch
import pandas as pd
%matplotlib inline

from Adaboost import Adaboost
```

```python
train_data = pd.read_csv('../../Data/phd/train_phd_data.csv')
test_data = pd.read_csv('../../Data/phd/test_phd_data.csv')
```

```python
train_tensor = torch.tensor(train_data.values, dtype=torch.float32)
train_data,train_label=train_tensor[:,:-1],train_tensor[:,-1]
test_tensor = torch.tensor(test_data.values, dtype=torch.float32)
test_data,test_label=test_tensor[:,:-1],test_tensor[:,-1]
```

# AdaBoost

```python
adaboost=Adaboost(T=10)
adaboost.fit(train_data,train_label.squeeze())

print(adaboost.predict(test_data))
print(test_label)
```
