```python
import torch
import numpy as np
%matplotlib inline
from sklearn.datasets import load_iris

from Softmax import SoftmaxRegression
```

```python
def create_data(train_rate: float = 0.8):
    iris = load_iris()

    X = torch.tensor(iris.data, dtype=torch.float32)   # (150, 4)
    y = torch.tensor(iris.target, dtype=torch.long)    # (150,)

    perm = torch.randperm(len(X))
    X = X[perm]
    y = y[perm]

    # one-hot 
    K = torch.max(y).item() + 1
    Y = torch.eye(K)[y]   # (150, K)

    train_size = int(train_rate * len(X))
    X_train = X[:train_size]
    Y_train = Y[:train_size]

    X_test = X[train_size:]
    Y_test = Y[train_size:]

    return X_train, Y_train, X_test, Y_test

X_train, Y_train, X_test, Y_test = create_data()
```

# Softmax

```python
softmax = SoftmaxRegression()

softmax.train(X=X_train, Y=Y_train)
_,preds = softmax.predict(X_test)
```

```python
y_true = torch.argmax(Y_test, dim=1)   # (30,)
mistakes = torch.sum(preds != y_true)
accuracy = 100 * (1 - mistakes.item() / len(preds))

print(f"Accuracy: {accuracy}%")
```
