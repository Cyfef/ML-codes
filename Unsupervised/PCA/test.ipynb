```python
import torch
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

from pca import PCA
```

```python
def create_data(n_samples_per_class=200):
    """
    Returns:
        X:  (3*n_samples_per_class, 2) 
    """
    # centers
    center1 = torch.tensor([2.0, 0.0])
    center2 = torch.tensor([0.0, 0.0])
    center3 = torch.tensor([-2.0, 0.0])

    # noise
    class1 = torch.randn(n_samples_per_class, 2) + center1
    class2 = torch.randn(n_samples_per_class, 2) + center2
    class3 = torch.randn(n_samples_per_class, 2) + center3

    # cat
    X = torch.cat([class1, class2, class3], dim=0)

    # scale,rotate
    theta = torch.tensor(torch.pi / 4)    # 45°
    scale = torch.diag(torch.tensor([1.2, 0.5]))
    rotate = torch.tensor([[torch.cos(theta), -torch.sin(theta)],
                           [torch.sin(theta),  torch.cos(theta)]])

    # transform
    X = X @ scale @ rotate.T
    return X

X=create_data()
```

# PCA

```python
pca_1=PCA(1)
pca_2=PCA(2)

Y=pca_2.reduce(X)
Z=pca_1.reduce(X)
```

```python
plt.figure(figsize=[15, 5])

# GT
plt.subplot(1, 3, 1)
plt.title("Ground Truth")
plt.xlim(-5, 5)
plt.ylim(-5, 5)
for cluster in X.reshape(3, -1, 2):
    plt.scatter(cluster[:, 0], cluster[:, 1], marker=".")

# K=2   
plt.subplot(1, 3, 2)
plt.title("PCA 2D")
plt.xlim(-5, 5)
plt.ylim(-5, 5)
for cluster in Y.reshape(3, -1, 2):
    plt.scatter(cluster[:, 0], cluster[:, 1], marker=".")

# K=3
plt.subplot(1, 3, 3)
plt.title("PCA 1D")
plt.xlim(-5, 5)
plt.ylim(-5, 5)
for cluster in Z.reshape(3, -1):
    plt.scatter(cluster, np.zeros_like(cluster), marker=".")

plt.tight_layout()
plt.show()
```
